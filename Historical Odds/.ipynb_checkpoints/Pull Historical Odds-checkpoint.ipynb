## Collect NBA Event IDS

```python
import requests
import sqlite3
from datetime import datetime, timedelta
```

```python
# Connect to (or create) a SQLite database file
conn = sqlite3.connect('event_ids.db')
cursor = conn.cursor()

# Create a table for events if it doesn't exist
cursor.execute('''
CREATE TABLE IF NOT EXISTS events (
    id TEXT PRIMARY KEY,
    sport_key TEXT,
    sport_title TEXT,
    commence_time TEXT,
    home_team TEXT,
    away_team TEXT
)
''')
conn.commit()
```

```python
API_KEY = 'a7d1378f0dea0dff37f2df1ab0769238'
SPORT = 'basketball_nba'  # Adjust sport key as needed
BASE_URL = f'https://api.the-odds-api.com/v4/historical/sports/{SPORT}/events'

# Start and end dates
start_date = datetime(2023, 5, 3)  # May 3, 2023
end_date = datetime.utcnow()       # Current UTC time

# Set interval for iteration (for example, 1 day)
interval = timedelta(days=1)

current_date = start_date
```

```python
while current_date <= end_date:
    date_str = current_date.strftime('%Y-%m-%dT%H:%M:%SZ')
    
    params = {
        'apiKey': API_KEY,
        'date': date_str,
        # Optional filters can go here
    }
    
    response = requests.get(BASE_URL, params=params)
    
    if response.status_code == 200:
        data = response.json()
        events = data.get('data', [])
        
        for event in events:
            # Prepare event data
            event_data = (
                event.get('id'),
                event.get('sport_key'),
                event.get('sport_title'),
                event.get('commence_time'),
                event.get('home_team'),
                event.get('away_team')
            )
            try:
                # Insert event into database
                cursor.execute('''
                    INSERT INTO events (id, sport_key, sport_title, commence_time, home_team, away_team)
                    VALUES (?, ?, ?, ?, ?, ?)
                ''', event_data)
            except sqlite3.IntegrityError:
                # Event ID already exists due to PRIMARY KEY constraint; skip insertion
                continue
        
        # Commit after inserting events for this timestamp
        conn.commit()
        print(f"Processed snapshot {date_str}: {len(events)} events.")
    else:
        print(f"Error on {date_str}: {response.status_code} - {response.text}")
    
    current_date += interval

print("Data collection complete.")
```

```python
# Close the database connection when done
conn.close()
```

### Pull Historical Odds by NBA Event IDS

```python
import requests
import sqlite3
from datetime import datetime
import time

def american_to_decimal(american):
    """Convert American odds to Decimal odds."""
    if american is None:
        return None
    return (american / 100) + 1 if american > 0 else (100 / abs(american)) + 1

# --- Configuration ---
API_KEY = '1aca707597d592e03a44d5762216606b'
SPORT = 'basketball_nba'
BASE_ODDS_URL = 'https://api.the-odds-api.com/v4/historical/sports'
REGIONS = 'us'
MARKETS = (
    'player_points,h2h_q1,player_assists,player_rebounds,player_points_rebounds_assists,'
    'player_points_rebounds,player_points_assists,player_rebounds_assists,player_points_alternate,'
    'player_rebounds_alternate,player_assists_alternate,player_points_rebounds_assists_alternate'
)

# --- Connect to Databases ---

# Connect to the database with stored event IDs
events_conn = sqlite3.connect('event_ids.db')
events_cursor = events_conn.cursor()

# Connect to (or create) a separate database to store the historical odds
odds_conn = sqlite3.connect('historical_player_odds.db')
odds_cursor = odds_conn.cursor()

# Create the odds table if it doesn't exist
odds_cursor.execute('''
CREATE TABLE IF NOT EXISTS odds (
    event_id TEXT,
    bookmaker TEXT,
    market TEXT,
    outcome_name TEXT,
    description TEXT,
    point REAL,
    american_odds REAL,
    decimal_odds REAL,
    snapshot_time TEXT,
    PRIMARY KEY (event_id, bookmaker, market, outcome_name, point)
)
''')
odds_conn.commit()

# --- Fetch Event IDs and Commence Times from event_ids.db ---
events_cursor.execute('SELECT id, commence_time FROM events')
event_rows = events_cursor.fetchall()
events_conn.close()

total_events = len(event_rows)
processed_events = 0
failed_events = 0

for event_id, commence_time in event_rows:
    processed_events += 1
    
    if not commence_time:
        print(f"[{processed_events}/{total_events}] Skipping event {event_id} due to missing commence_time.")
        failed_events += 1
        continue

    # The historical odds endpoint requires the 'date' param to be the original
    # commence_time (snapshot time) in ISO8601 format, or as required by the API.
    odds_date = commence_time

    # Construct the URL for historical odds
    url = f"{BASE_ODDS_URL}/{SPORT}/events/{event_id}/odds"
    params = {
        'apiKey': API_KEY,
        'date': odds_date,
        'regions': REGIONS,
        'markets': MARKETS,
        'dateFormat': 'iso',
        'oddsFormat': 'american'
    }

    try:
        response = requests.get(url, params=params)
    except requests.exceptions.RequestException as e:
        print(f"[{processed_events}/{total_events}] Network error for event {event_id}: {e}")
        failed_events += 1
        continue

    if response.status_code == 200:
        odds_data = response.json()

        # The API may return a "timestamp" field or "last_update" field.
        # If not present, fall back to the commence_time as a snapshot.
        snapshot_time = odds_data.get('timestamp', odds_date)

        data = odds_data.get('data', {})
        bookmakers = data.get('bookmakers', [])

        for bookmaker in bookmakers:
            bookmaker_key = bookmaker.get('key')
            markets_list = bookmaker.get('markets', [])
            
            for market in markets_list:
                market_key = market.get('key')
                outcomes = market.get('outcomes', [])
                
                for outcome in outcomes:
                    outcome_name = outcome.get('name')
                    description = outcome.get('description')
                    point = outcome.get('point')
                    american_odds = outcome.get('price')
                    decimal_odds = american_to_decimal(american_odds)

                    odds_record = (
                        event_id,
                        bookmaker_key,
                        market_key,
                        outcome_name,
                        description,
                        point,
                        american_odds,
                        decimal_odds,
                        snapshot_time
                    )

                    try:
                        odds_cursor.execute('''
                            INSERT INTO odds 
                            (event_id, bookmaker, market, outcome_name, description, point, american_odds, decimal_odds, snapshot_time)
                            VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?)
                        ''', odds_record)
                    except sqlite3.IntegrityError:
                        # The PRIMARY KEY constraint (duplicate record) triggers an IntegrityError
                        pass

        odds_conn.commit()
        print(f"[{processed_events}/{total_events}] Stored odds for event {event_id}")

    elif response.status_code == 404:
        error_response = response.json()
        message = error_response.get('message', 'No message provided.')
        print(f"[{processed_events}/{total_events}] Not found for event {event_id}: {response.status_code} - {message}")
        failed_events += 1

    else:
        error_response = response.json()
        message = error_response.get('message', 'No message provided.')
        print(f"[{processed_events}/{total_events}] Error for event {event_id}: {response.status_code} - {message}")
        failed_events += 1

    # Optional: a small delay to avoid hammering the API or hitting rate limits
    time.sleep(0.5)

# --- Final Output ---
print(f"\nData collection complete. Processed {processed_events} events with {failed_events} failures.")

# Close the odds database connection
odds_conn.close()
```
